```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset("dipawidia/ecommerce-product-reviews-sentiment")
df = dataset['train'].to_pandas()

print("=== JUMLAH DATA & KOLOM ===")
print(f"Jumlah baris (data): {df.shape[0]}")
print(f"Jumlah kolom: {df.shape[1]}")
print(f"Nama kolom: {list(df.columns)}\n")

print("=== DISTRIBUSI LABEL SENTIMEN ===")
kolom_label = df.columns[1] 
print(df[kolom_label].value_counts())

print("\n=== CONTOH 5 DATA TERATAS ===")
print(df.head())
```

```python
import re

def cleansing(text):
    if pd.isna(text):
        return ""
    text = str(text).lower() # Case folding (huruf kecil)
    text = re.sub(r'[^a-zA-Z\s]', '', text) # Menghapus angka dan simbol
    text = re.sub(r'\s+', ' ', text).strip() # Menghapus spasi berlebih
    return text

df['cleaned_review'] = df['review'].apply(cleansing)

print("=== CONTOH HASIL TEXT CLEANING ===")
print(df[['review', 'cleaned_review']].head())
```

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')


df['tokens'] = df['cleaned_review'].apply(word_tokenize)

print("=== CONTOH HASIL TOKENISASI ===")
print(df[['cleaned_review', 'tokens']].head())
```

```python
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

stop_factory = StopWordRemoverFactory()
stopwords_indonesia = stop_factory.get_stop_words()
stopword_set = set(stopwords_indonesia)

df['tokens_stopword'] = df['tokens'].apply(lambda x: [word for word in x if word not in stopword_set])

print("=== CONTOH HASIL STOPWORD REMOVAL ===")
print(df[['tokens', 'tokens_stopword']].head())
```

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

factory = StemmerFactory()
stemmer = factory.create_stemmer()

def stem_tokens(tokens):
    return [stemmer.stem(word) for word in tokens]

df['tokens_stemmed'] = df['tokens_stopword'].apply(stem_tokens)

print("=== CONTOH HASIL STEMMING ===")
print(df[['tokens_stopword', 'tokens_stemmed']].head())
```
